```python
#To read bronze table
bronze_df =  spark.table("workspace.default.bronze_hhs_healthcare_breaches")
display(bronze_df)
```

```python
print(bronze_df.columns)
```

```python
#Creating Silver Table
from pyspark.sql.functions import col, trim, lower, upper, year, month, when, lit

silver_df = (
    bronze_df
    .withColumn("covered_entity_name", trim(col("covered_entity_name")))
    .withColumn("state", upper(trim(col("state"))))
    .withColumn("covered_entity_type", trim(col("covered_entity_type")))
    .withColumn("type_of_breach", trim(col("type_of_breach")))
    .withColumn("location_of_breached_information", trim(col("location_of_breached_information")))
    .withColumn("individuals_affected", col("individuals_affected").cast("int"))
    .withColumn("breach_year", year(col("breach_submission_date")))
    .withColumn("breach_month", month(col("breach_submission_date")))
)

display(silver_df)
```

```python
#Add Risk level
silver_df = (
    silver_df
    .withColumn(
        "risk_level",
        when(
            (col("individuals_affected") >= 100000) |
            (lower(col("type_of_breach")).contains("hacking")),
            "High"
        )
        .when(
            (col("individuals_affected") >= 10000) |
            (lower(col("type_of_breach")).contains("unauthorized")),
            "Medium"
        )
        .otherwise("Low")
    )
)

display(silver_df)
```

```python
print(silver_df.columns)
```

```python
# Find remaining HHS exported weird column
weird_cols = [c for c in silver_df.columns if c.startswith("javax.faces.component.UIPanel")]

print(weird_cols)
```

```python
if len(weird_cols) > 0:
    silver_df = silver_df.withColumnRenamed(weird_cols[0], "business_associate_present")

display(silver_df)
```

```python
silver_df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable(
    "workspace.default.silver_hhs_healthcare_breaches"
)
```
